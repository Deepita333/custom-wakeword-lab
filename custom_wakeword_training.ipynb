## 1. Runtime preflight

Run this first. It reports the Python version, available disk space, and memory without downloading anything.

```python
import os
import sys
import shutil
import platform

print("Python:", sys.version)
print("Platform:", platform.platform())
print("Working directory:", os.getcwd())

disk = shutil.disk_usage("/content" if os.path.exists("/content") else os.getcwd())
print(f"Disk total: {disk.total / (1024**3):.2f} GiB")
print(f"Disk free:  {disk.free / (1024**3):.2f} GiB")

try:
    import psutil
    print(f"RAM total:  {psutil.virtual_memory().total / (1024**3):.2f} GiB")
    print(f"RAM free:   {psutil.virtual_memory().available / (1024**3):.2f} GiB")
except ImportError:
    print("RAM info unavailable (psutil is not installed).")
```
